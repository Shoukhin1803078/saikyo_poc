```python
# Intent Classifier Function (LangChain ChatOllama)

import json
from langchain_ollama import ChatOllama

from pydantic import BaseModel
from typing import List

class IntentResponse(BaseModel):
    matched_intents: List[str]



ALL_INTENTS = ["identity", "work_content", "requirements", "compensation", "benefits", "logistics", "hiring_flow"]



# Initialize the LLM
llm = ChatOllama(
    model="qwen2.5:7b",  # your local model
    temperature=0
)

def classify_intent(user_query: str) -> List[str]:
    """
    Classify user_query into one or more predefined intents using ChatOllama.
    Returns a list of matched intents.
    """
    messages = [
        (
            "system",
            "You are a semantic intent classifier. "
            "Your task is to select which predefined intent(s) match a user query."
        ),
        (
            "human",
            f"User query: \"{user_query}\"\n"
            f"Predefined intents: {ALL_INTENTS}\n"
            "Pick all relevant intents and return them as a JSON array ONLY. Example: [\"work_content\", \"requirements\"]"
        )
    ]

    ai_msg = llm.invoke(messages)
    print(ai_msg)

    try:
        intents = json.loads(ai_msg)
        return [i for i in intents if i in ALL_INTENTS]
    except Exception:
        return []


user_query = "私はソフトウェアエンジニアです。AI/ML、Pythonの経験が4年あります。ダッカの勤務地で私の経験に基づいて仕事を提案してください。"

matched_intents = classify_intent(user_query)

print(matched_intents)


```

```python
from typing import List, Literal
from pydantic import BaseModel
from langchain_ollama import ChatOllama

# Predefined intents
ALL_INTENTS = ["identity", "work_content", "requirements", "compensation", "benefits", "logistics", "hiring_flow"]

# Pydantic model with Literal for strict validation
class IntentResponse(BaseModel):
    matched_intents: List[Literal[
        "identity", "work_content", "requirements", "compensation", "benefits", "logistics", "hiring_flow"
    ]]

# Initialize LLM
llm = ChatOllama(model="qwen2.5:7b", temperature=0)

# Wrap LLM with structured output
llm_structured = llm.with_structured_output(IntentResponse)

# Function to classify intent
def classify_intent(user_query: str) -> IntentResponse:
    prompt = (
        f"Classify this user query into relevant intents from the list: {ALL_INTENTS}\n"
        f"User query: {user_query}\n"
        "Return the intents only. Do not include any text outside the list."
    )

    llm_response=llm_structured.invoke(prompt)
    return llm_response

# Example usage
user_query = "私はソフトウェアエンジニアです。AI/ML、Pythonの経験が4年あります。ダッカの勤務地で私の経験に基づいて仕事を提案してください。"
response = classify_intent(user_query)

print(response)  
print(response.matched_intents)  # ['work_content', 'requirements']
```
